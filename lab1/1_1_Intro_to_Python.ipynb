4. Classes

```python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def grade(self):
        # return letter grade based on numeric score
        if self.score >= 90:
            return 'A'
        elif self.score >= 80:
            return 'B'
        elif self.score >= 70:
            return 'C'
        else:
            return 'F'

s = Student('Dana', 85)
print(s.name, s.grade())  # Dana B
```
