Завдання 1

```python
import pandas as pd

url = 'https://raw.githubusercontent.com/smir-nat/IAD-Practicals/refs/heads/main/lab1/cybersecurity_lab1.csv'
df = pd.read_csv(url)

print(f"Розмір датасета: {df.shape[0]} рядків та {df.shape[1]} колонок\n")
print("Назви колонок:")
print(df.columns.tolist())
print("\nТипи даних у кожній колонці:")
print(df.dtypes)
print("\nРозподіл класів (цільова змінна 'label'):")
print(df['label'].value_counts())

print("\nДетальний розподіл типів атак (колонка 'attack_type'):")
print(df['attack_type'].value_counts())
```

Завдання 2

```python
from sklearn.preprocessing import LabelEncoder

df['url'] = df['url'].fillna('Unknown')
print("Перевірка пропусків після заповнення:")
print(df.isnull().sum())

cols_to_drop = ['timestamp', 'src_ip', 'dst_ip', 'url', 'attack_type']
df_clean = df.drop(columns=cols_to_drop)

le = LabelEncoder()
df_clean['protocol'] = le.fit_transform(df_clean['protocol'])
df_clean['user_agent'] = le.fit_transform(df_clean['user_agent'])

df_clean['is_internal_traffic'] = df_clean['is_internal_traffic'].astype(int)

print("\nОновлений датасет після обробки (перші 5 рядків):")
display(df_clean.head())

print("\nТипи даних тепер:")
print(df_clean.dtypes)
```

Завдання 3

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

# 1. Heatmap кореляцій
plt.figure(figsize=(9, 7))
corr = df_clean.corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title("Теплова карта (Heatmap) кореляцій ознак", fontsize=14)
plt.show()

# 2. Гістограми розподілу найбільш показових ознак
fig, axes = plt.subplots(1, 3, figsize=(16, 4))
sns.histplot(df_clean['src_port'], kde=True, ax=axes[0], color='steelblue')
axes[0].set_title('Розподіл src_port')

sns.histplot(df_clean['bytes_sent'], kde=True, ax=axes[1], color='crimson')
axes[1].set_title('Розподіл bytes_sent')

sns.histplot(df_clean['dst_port'], kde=True, ax=axes[2], color='seagreen')
axes[2].set_title('Розподіл dst_port')
plt.tight_layout()
plt.show()

# 3. Boxplot-и ознак відносно цільової змінної (label)
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Boxplot для bytes_sent
sns.boxplot(x='label', y='bytes_sent', data=df_clean, ax=axes[0], palette='Set2')
axes[0].set_yscale('log')
axes[0].set_title('Розподіл bytes_sent за класами (log-scale)')
axes[0].set_xlabel('Клас (0 - Benign, 1 - Attack)')

# Boxplot для src_port
sns.boxplot(x='label', y='src_port', data=df_clean, ax=axes[1], palette='Set2')
axes[1].set_title('Розподіл src_port за класами')
axes[1].set_xlabel('Клас (0 - Benign, 1 - Attack)')

plt.tight_layout()
plt.show()
```
